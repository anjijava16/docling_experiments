```python
from docling.document_converter import DocumentConverter
import base64
from io import BytesIO

# Convert the document
source = "https://arxiv.org/pdf/2408.09869"
converter = DocumentConverter()
result = converter.convert(source)

# Get the markdown with placeholders
markdown_text = result.document.export_to_markdown()

# Replace each picture placeholder with a base64-encoded image
for i, picture in enumerate(result.document.pictures):
    # Get the image as PIL Image
    image = picture.get_image()   # Returns a PIL Image
    
    # Convert PIL Image to bytes (e.g., PNG format)
    buffer = BytesIO()
    image.save(buffer, format="PNG")
    img_bytes = buffer.getvalue()
    
    # Encode to base64
    base64_str = base64.b64encode(img_bytes).decode("utf-8")
    data_uri = f"data:image/png;base64,{base64_str}"
    
    # Replace the placeholder (assuming the placeholder is "![](image_placeholder)")
    # You might need to adjust the placeholder pattern based on actual output.
    placeholder = f"![Image {i}]"
    markdown_text = markdown_text.replace(placeholder, f"![Image {i}]({data_uri})")

# Now markdown_text contains inline base64 images
print(markdown_text)
```

```python
from docling.document_converter import DocumentConverter
import base64
from io import BytesIO
from PIL import Image as PILImage  # ensure Pillow is imported

source = "https://arxiv.org/pdf/2408.09869"
converter = DocumentConverter()
result = converter.convert(source)

markdown_text = result.document.export_to_markdown()

for i, picture in enumerate(result.document.pictures):
    img_obj = picture.get_image(result.document)
    
    # Debug: see what we got
    print(f"Type of image {i}: {type(img_obj)}")
    
    # Convert to a PIL Image if needed
    if hasattr(img_obj, 'pil'):          # Docling wrapper with .pil attribute
        pil_image = img_obj.pil
    elif hasattr(img_obj, 'to_pil'):     # Some method to convert
        pil_image = img_obj.to_pil()
    elif isinstance(img_obj, PILImage.Image):
        pil_image = img_obj
    else:
        # Try to convert via numpy if it's array-like
        import numpy as np
        if isinstance(img_obj, np.ndarray):
            pil_image = PILImage.fromarray(img_obj)
        else:
            raise TypeError(f"Unsupported image type: {type(img_obj)}")
    
    buffer = BytesIO()
    pil_image.save(buffer, format="PNG")
    img_bytes = buffer.getvalue()
    base64_str = base64.b64encode(img_bytes).decode("utf-8")
    data_uri = f"data:image/png;base64,{base64_str}"
    
    # Replace placeholder (adjust as needed)
    placeholder = "![](image_placeholder)"
    markdown_text = markdown_text.replace(placeholder, f"![Image {i}]({data_uri})", 1)

print(markdown_text)
```

```python
from docling.document_converter import DocumentConverter
import base64
import io

source = "https://arxiv.org/pdf/2408.09869"

converter = DocumentConverter()
result = converter.convert(source)
doc = result.document

markdown = doc.export_to_markdown()

for i, pic in enumerate(doc.pictures):

    image = None

    # ---- Try different possible fields depending on Docling version ----
    if hasattr(pic, "image") and pic.image is not None:
        image = pic.image

    elif hasattr(pic, "pil_image") and pic.pil_image is not None:
        image = pic.pil_image

    elif hasattr(pic, "data") and pic.data is not None:
        # raw bytes → PIL Image
        from PIL import Image
        image = Image.open(io.BytesIO(pic.data))

    # ---- Skip if still None ----
    if image is None:
        print(f"Skipping picture {i} (no image data)")
        continue

    # Convert to base64
    buffer = io.BytesIO()
    image.save(buffer, format="PNG")
    b64 = base64.b64encode(buffer.getvalue()).decode("utf-8")

    placeholder = f"![image_{i}](image_{i}.png)"

    inline_img = f"![image_{i}](data:image/png;base64,{b64})"
    print(inline_img)

    markdown = markdown.replace(placeholder, inline_img)

print(markdown)
```

```python
from docling.document_converter import DocumentConverter, PdfFormatOption
from docling.datamodel.base_models import InputFormat
from docling.datamodel.pipeline_options import PdfPipelineOptions
import base64
import io

source = "https://arxiv.org/pdf/2408.09869"

options = PdfPipelineOptions()
options.do_ocr = False
options.render_pages = True
options.render_figures = True
options.generate_picture_images = True

converter = DocumentConverter(
    format_options={
        InputFormat.PDF: PdfFormatOption(
            pipeline_options=options
        )
    }
)

result = converter.convert(source)
doc = result.document

markdown = doc.export_to_markdown()

# ---- Convert rendered figures to base64 ----
for i, pic in enumerate(doc.pictures):

    if pic.image is None:
        print(f"No rendered image for picture {i}")
        continue

    # In v2, pic.image is a PIL Image if rendering is enabled
    image = pic.image

    buffer = io.BytesIO()
    image.save(buffer, format="PNG")
    b64 = base64.b64encode(buffer.getvalue()).decode()

    placeholder = f"![image_{i}](image_{i}.png)"
    inline_img = f"![image_{i}](data:image/png;base64,{b64})"

    markdown = markdown.replace(placeholder, inline_img)

print(markdown)


```
